```python
'''Implement Reinforcement Learning using an example of a maze environment that the 
agent needs to explore.'''
import numpy as np

# Define maze (0 = Path, 1 = Wall)
maze = np.array([
    [0, 0, 0, 1],
    [1, 1, 0, 1],
    [0, 0, 0, 0],
    [0, 1, 1, 0]
])

# Start and Goal positions
start = (0, 0)
goal = (3, 3)

# Actions: Up, Down, Left, Right
actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

# Initialize Q-table
q_table = np.zeros((4, 4, 4))

# Learning parameters
alpha = 0.1
gamma = 0.9
epsilon = 0.1
episodes = 1000

# Q-Learning
for episode in range(episodes):
    state = start

    for step in range(100):
        if np.random.rand() < epsilon:
            action = np.random.randint(4)
        else:
            action = np.argmax(q_table[state[0], state[1]])

        new_row = state[0] + actions[action][0]
        new_col = state[1] + actions[action][1]

        # Check valid movement
        if (0 <= new_row < 4 and 0 <= new_col < 4 and
                maze[new_row, new_col] == 0):
            next_state = (new_row, new_col)
        else:
            next_state = state

        # Reward
        if next_state == goal:
            reward = 100
        else:
            reward = -1

        # Update Q-table
        old_value = q_table[state[0], state[1], action]
        next_max = np.max(q_table[next_state[0], next_state[1]])

        q_table[state[0], state[1], action] = (
            old_value + alpha * (reward + gamma * next_max - old_value)
        )

        state = next_state

        if state == goal:
            break

# Display learned path
state = start
path = [state]

for step in range(20):
    if state == goal:
        break

    action = np.argmax(q_table[state[0], state[1]])
    state = (state[0] + actions[action][0],
             state[1] + actions[action][1])

    path.append(state)

    if maze[state[0], state[1]] == 1:
        break

print("Maze:")
print(maze)
print("\nLearned Path:")
print(path)

if path[-1] == goal:
    print("\nGoal Reached!")
else:
    print("\nGoal Not Reached!")
```
